## 3. Cargar tablas a SQLite

```python
conn = sqlite3.connect(":memory:")

estudiantes.to_sql("estudiantes", conn, index=False, if_exists="replace")
cursos.to_sql("cursos", conn, index=False, if_exists="replace")
inscripciones.to_sql("inscripciones", conn, index=False, if_exists="replace")
pagos.to_sql("pagos", conn, index=False, if_exists="replace")

def run_sql(query):
    return pd.read_sql_query(query, conn)
```

### Punto 4. Agregación simple

Calcula:
- número total de cursos
- precio promedio
- precio mínimo
- precio máximo

Usa la tabla `cursos`.

**Entregable:**

```python
# Punto 4 - SQL
Query4 = """
SELECT COUNT(*) AS total_cursos,
       AVG(precio) AS precio_promedio,
       MIN(precio) AS precio_minimo,
       MAX(precio) AS precio_maximo
FROM cursos
"""
run_sql(Query4)
```

### Punto 5. GROUP BY

### Punto 6. JOIN entre tablas

Muestra:
- nombre del estudiante
- nombre del curso
- estado de la inscripción

Usa `estudiantes`, `inscripciones` y `cursos`.

**Entregable:**
- SQL
- pandas

```python
# Punto 6 - SQL
Query6 = """
SELECT e.nombre AS nombre_estudiante, c.nombre_curso, i.estado
FROM estudiantes e
JOIN inscripciones i ON e.estudiante_id = i.estudiante_id
JOIN cursos c ON i.curso_id = c.curso_id
"""
run_sql(Query6)
```

```python
# Punto 6 - pandas
df_estudiantes.merge(inscripciones, on="estudiante_id").merge(cursos, on="curso_id")[["nombre", "nombre_curso", "estado"]]
```

### Punto 7. LEFT JOIN

Trae todos los estudiantes, incluso si no tuvieran inscripciones, mostrando:
- `estudiante_id`
- `nombre`
- `inscripcion_id`

**Entregable:**
- SQL
- pandas

> Pista: en este dataset puede que todos tengan al menos una inscripción, pero igual debes practicar el `LEFT JOIN`.

```python
# Punto 7 - SQL
Query7 = """
SELECT e.estudiante_id, e.nombre, i.inscripcion_id
FROM estudiantes e
LEFT JOIN inscripciones i ON e.estudiante_id = i.estudiante_id
"""
run_sql(Query7)
```

```python
# Punto 7 - pandas
df_estudiantes.merge(inscripciones, on="estudiante_id", how="left")[["estudiante_id", "nombre", "inscripcion_id"]]
```

### Punto 8. HAVING

Calcula cuántas inscripciones tiene cada estudiante y muestra solo aquellos con **2 o más**.

Debes mostrar:
- `estudiante_id`
- `nombre`
- `n_inscripciones`

**Entregable:**
- SQL
- pandas

```python
# Punto 8 - SQL
Query8 = """
SELECT e.estudiante_id, e.nombre, COUNT(i.inscripcion_id) AS n_inscripciones
FROM estudiantes e
LEFT JOIN inscripciones i ON e.estudiante_id = i.estudiante_id
GROUP BY e.estudiante_id, e.nombre
HAVING n_inscripciones >= 2
"""
run_sql(Query8)
```

```python
#Punto 8 - Pandas
table = estudiantes.merge(inscripciones, on="estudiante_id", how="left").groupby(["estudiante_id", "nombre"]).size().reset_index(name="n_inscripciones")
table = table[table["n_inscripciones"] >= 2]
table
```

### Punto 9. CASE WHEN / segmentación

Clasifica los cursos según su precio:
- `Bajo` si el precio es menor a 120
- `Medio` si el precio está entre 120 y 180 inclusive
- `Alto` si el precio es mayor a 180

Debes mostrar:
- `nombre_curso`
- `precio`
- `segmento_precio`

**Entregable:**
- SQL
- pandas

```python
# Punto 9 - SQL
Query8 = """
SELECT nombre_curso, precio,
        CASE
        WHEN precio < 120 THEN 'Bajo'
        WHEN precio <= 180 THEN 'Medio'
        ELSE 'Alto'
    END AS segmento_precio
FROM cursos
"""
run_sql(Query8)
```

```python
# Punto 9 - pandas
cursos["segmento_precio"] = pd.cut(cursos["precio"], bins=[0, 120, 180, float("inf")], labels=["Bajo", "Medio", "Alto"])
cursos[["nombre_curso", "precio", "segmento_precio"]]
```

### Punto 10. Ejercicio final integrador

Calcula el total pagado por categoría de curso.

Debes:
- unir `cursos`, `inscripciones` y `pagos`
- agrupar por `categoria`
- sumar `monto`
- ordenar de mayor a menor

**Entregable:**
- SQL
- pandas

```python
# Punto 10 - SQL
Query10 = """
SELECT c.categoria, SUM(p.monto) AS total_pagado
FROM cursos c
JOIN inscripciones i ON c.curso_id = i.curso_id
JOIN pagos p ON i.inscripcion_id = p.inscripcion_id
GROUP BY c.categoria
"""
run_sql(Query10)
```

```python
# Punto 10 - pandas
Merge = cursos.merge(inscripciones, on="curso_id").merge(pagos, on="inscripcion_id")
Merge.groupby("categoria")["monto"].sum().sort_values(ascending=False)
Merge
```
